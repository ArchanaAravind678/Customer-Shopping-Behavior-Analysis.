```python
# create age group

import pandas as pd
import urllib.parse
df=pd.read_csv('customer_shopping_behavior.csv')
# create age group
labels=['18–25','26–40','41–60','61+']
df['age_group']=pd.qcut(df['Age'],q=4,labels=labels)
from sqlalchemy import create_engine


# 2. TRANSFORM FIELD NAMES: Convert all columns to lowercase and replace spaces with underscores
df.columns = df.columns.str.lower().str.replace(' ', '_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})



# 2. Define the name of your new Excel file
excel_filename = 'customer_shopping_behavior.xlsx'

# 3. Export the DataFrame to an Excel file
df.to_excel(excel_filename, index=False, sheet_name='customer_shopping_behavior')

# 4. Print confirmation (Fixed variable name inside the brackets here)
print(f"Data successfully saved to Excel file: {excel_filename}")



# 2. Connect to a local SQLite database (it will create the file if it doesn't exist)
# 3. Connection variables
USER = "root"
PASSWORD = urllib.parse.quote_plus("Archana@123")
HOST = "127.0.0.1"
DATABASE = "customer_behaviour"

# 4. Initialize MySQL engine connection
engine = create_engine(f"mysql+mysqlconnector://{USER}:{PASSWORD}@{HOST}/{DATABASE}")

# 5. Load data into the 'customer' table
# Set if_exists='append' if the table already exists and you want to add rows
# Set if_exists='replace' if you want to drop the old structure and create it brand new
df.to_sql('customer', con=engine, if_exists='replace', index=False)

print("Data successfully mapped and loaded into the 'customer' table!")
```
